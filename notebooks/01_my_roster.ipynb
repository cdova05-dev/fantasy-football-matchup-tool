```python
my_player_info = (
    players
    .filter(
        players["display_name"].is_in(MY_PLAYERS)
        & players["position"].is_in(["RB", "WR", "TE"])
    )
    .select([
        "gsis_id",
        "display_name",
        "position",
        "latest_team"
    ])
)

my_player_info
```

```python
schedule = nfl.load_schedules()

print(schedule.shape)

for col in schedule.columns:
    print(col)
```

```python
CURRENT_SEASON = nfl.get_current_season()
CURRENT_WEEK = nfl.get_current_week()

print("Season:", CURRENT_SEASON)
print("Week:", CURRENT_WEEK)

week_schedule = (
    schedule.filter(
        (schedule['season'] == CURRENT_SEASON)
        & (schedule['week'] == CURRENT_WEEK)
        & (schedule['game_type'] == 'REG')
    )
    .select([
        'week',
        'gameday',
        'gametime',
        'away_team',
        'home_team'
    ])
)

week_schedule
```

```python
team_stats = nfl.load_team_stats()

print(team_stats.shape)

for col in team_stats.columns:
    print(col)
```

```python
print(
    team_stats
    .select("week")
    .unique()
    .sort("week")
)
```

```python
import polars as pl
```

```python
away_rows = week_schedule.select([
    pl.col("away_team").alias("team"),
    pl.col("home_team").alias("opponent"),
    pl.lit("A").alias("home_away"),
    "gameday",
    "gametime"
])

home_rows = week_schedule.select([
    pl.col("home_team").alias("team"),
    pl.col("away_team").alias("opponent"),
    pl.lit("H").alias("home_away"),
    "gameday",
    "gametime"
])

team_matchups = pl.concat([away_rows, home_rows])

team_matchups
```

```python
my_roster_matchups = (
    my_player_info
    .join(
        team_matchups,
        left_on="latest_team",
        right_on="team",
        how="left"
    )
)

my_roster_matchups
```

```python
defense_check = (
    team_stats
    .filter(
        (team_stats["season"] == CURRENT_SEASON)
        & (team_stats["season_type"] == "REG")
        & (team_stats["week"] < CURRENT_WEEK)
    )
    .select([
        "week",
        "team",
        "opponent_team",
        "passing_yards",
        "passing_tds",
        "rushing_yards",
        "rushing_tds",
        "attempts",
        "carries",
        "sacks_suffered",
        "sack_yards_lost"
    ])
)

defense_check
```

DEFENSE

```python
defense_efficiency = (
    defense_check
    .group_by("opponent_team")
    .agg([
        pl.col("rushing_yards").sum().alias("rush_yards_allowed_total"),
        pl.col("carries").sum().alias("rush_attempts_faced"),

        pl.col("passing_yards").sum().alias("pass_yards_allowed_total"),
        pl.col("sack_yards_lost").sum().alias("sack_yards_lost_total"),
        pl.col("attempts").sum().alias("pass_attempts_faced"),
        pl.col("sacks_suffered").sum().alias("sacks_faced"),

        pl.col("rushing_yards").sum().alias("total_rush_yards"),
        pl.col("passing_yards").sum().alias("total_pass_yards")
    ])
    .with_columns([
        (
            pl.col("rush_yards_allowed_total")
            / pl.col("rush_attempts_faced")
        ).alias("rush_ypc_allowed"),

        (
            (
                pl.col("pass_yards_allowed_total")
                - pl.col("sack_yards_lost_total")
            )
            /
            (
                pl.col("pass_attempts_faced")
                + pl.col("sacks_faced")
            )
        ).alias("net_pass_yds_per_dropback"),

        (
            (
                pl.col("total_pass_yards")
                - pl.col("sack_yards_lost_total")
                + pl.col("total_rush_yards")
            )
            /
            (
                pl.col("pass_attempts_faced")
                + pl.col("sacks_faced")
                + pl.col("rush_attempts_faced")
            )
        ).alias("yards_per_play_allowed")
    ])
    .rename({
        "opponent_team": "defense_team"
    })
    .select([
        "defense_team",
        "rush_ypc_allowed",
        "net_pass_yds_per_dropback",
        "yards_per_play_allowed"
    ])
    .with_columns([
        pl.col("rush_ypc_allowed").round(2),
        pl.col("net_pass_yds_per_dropback").round(2),
        pl.col("yards_per_play_allowed").round(2)
    ])
)

defense_efficiency
```

```python
defense_efficiency_check = (
    defense_check
    .with_columns([
        (
            pl.col("rushing_yards") / pl.col("carries")
        ).alias("rush_ypc_allowed"),

        (
            (pl.col("passing_yards") - pl.col("sack_yards_lost"))
            /
            (pl.col("attempts") + pl.col("sacks_suffered"))
        ).alias("net_pass_ypd_allowed"),

        (
            (
                pl.col("passing_yards")
                - pl.col("sack_yards_lost")
                + pl.col("rushing_yards")
            )
            /
            (
                pl.col("attempts")
                + pl.col("sacks_suffered")
                + pl.col("carries")
            )
        ).alias("yards_per_play_allowed")
    ])
)

defense_efficiency_check.select([
    "week",
    "team",
    "opponent_team",
    "rush_ypc_allowed",
    "net_pass_ypd_allowed",
    "yards_per_play_allowed"
])
```

```python
defense_stats = (
    defense_check
    .group_by("opponent_team")
    .agg([
        pl.col("passing_yards")
            .mean()
            .alias("pass_yards_allowed"),

        pl.col("passing_tds")
            .sum()
            .alias("pass_tds_allowed"),

        pl.col("rushing_yards")
            .mean()
            .alias("rush_yards_allowed"),

        pl.col("rushing_tds")
            .sum()
            .alias("rush_tds_allowed")
    ])
    .rename({
        "opponent_team": "defense_team"
    })
)

defense_stats
```

```python
final_board = (
    my_roster_matchups
    .join(
        defense_stats,
        left_on="opponent",
        right_on="defense_team",
        how="left"
    )
)

final_board
```

```python
player_baseline = (
    player_stats
    .filter(
        player_stats["player_display_name"].is_in(MY_PLAYERS)
        & (player_stats["week"] < CURRENT_WEEK)
        & player_stats["position"].is_in(["RB", "WR", "TE"])
    )
    .select([
        "player_id",
        "player_display_name",
        "position",
        "week",
        "team",
        "rushing_yards",
        "rushing_tds",
        "targets",
        "receptions",
        "receiving_yards",
        "receiving_tds"
    ])
)

player_baseline
```

```python
player_summary = (
    player_baseline
    .group_by([
        "player_id",
        "player_display_name",
        "position",
        "team"
    ])
    .agg([
        pl.len().alias("games_played"),

        pl.col("targets").mean().alias("targets_per_game"),
        pl.col("receptions").mean().alias("receptions_per_game"),
        pl.col("receiving_yards").mean().alias("receiving_yards_per_game"),

        pl.col("rushing_yards").mean().alias("rushing_yards_per_game"),

        pl.col("rushing_tds").sum().alias("rushing_tds"),
        pl.col("receiving_tds").sum().alias("receiving_tds")
    ])
)

player_summary
```

```python
final_board = (
    final_board
    .join(
        player_summary,
        left_on="gsis_id",
        right_on="player_id",
        how="left"
    )
)

final_board
```

```python
league_efficiency = defense_efficiency.select([
    pl.col("rush_ypc_allowed").mean().alias("league_rush_ypc"),
    pl.col("net_pass_yds_per_dropback").mean().alias("league_net_pass_ypd"),
    pl.col("yards_per_play_allowed").mean().alias("league_ypp")
])

league_efficiency
```

```python
LEAGUE_RUSH_YPC = league_efficiency["league_rush_ypc"][0]
LEAGUE_NET_PASS_YPD = league_efficiency["league_net_pass_ypd"][0]
LEAGUE_YPP = league_efficiency["league_ypp"][0]

print("League Rush YPC:", round(LEAGUE_RUSH_YPC, 2))
print("League Net Pass Yds/Dropback:", round(LEAGUE_NET_PASS_YPD, 2))
print("League Yards/Play:", round(LEAGUE_YPP, 2))
```

```python
defense_ranks = (
    defense_efficiency
    .with_columns([
        pl.col("rush_ypc_allowed")
          .rank("ordinal")
          .alias("rush_def_rank"),

        pl.col("net_pass_yds_per_dropback")
          .rank("ordinal")
          .alias("pass_def_rank")
    ])
    .select([
        "defense_team",
        "rush_def_rank",
        "pass_def_rank"
    ])
)

defense_ranks
```

```python
matchup_board = (
    matchup_board
    .join(
        defense_ranks,
        left_on="opponent",
        right_on="defense_team",
        how="left"
    )
)
```

```python
matchup_board = (
    matchup_board
    .join(
        defense_efficiency,
        left_on="opponent",
        right_on="defense_team",
        how="left"
    )
)

matchup_board
```

```python
matchup_board = matchup_board.with_columns([
    pl.when(pl.col("position") == "RB")
      .then(pl.col("rush_ypc_allowed") - LEAGUE_RUSH_YPC)
      .otherwise(
          pl.col("net_pass_yds_per_dropback") - LEAGUE_NET_PASS_YPD
      )
      .round(2)
      .alias("efficiency_vs_league")
])

matchup_board
```

```python
matchup_board = matchup_board.with_columns(
    pl.col("efficiency_vs_league")
    .map_elements(
        lambda x: f"{x:+.2f}" if x is not None else "BYE",
        return_dtype=pl.Utf8
    )
    .alias("efficiency_vs_league_display")
)

matchup_board
```

```python
matchup_board = matchup_board.with_columns([
    pl.when(pl.col("week_status") == "BYE WEEK")
      .then(pl.lit("BYE"))

      .when(pl.col("efficiency_vs_league") >= 0.50)
      .then(pl.lit("Favorable"))

      .when(pl.col("efficiency_vs_league") <= -0.50)
      .then(pl.lit("Tough"))

      .otherwise(pl.lit("Neutral"))
      .alias("matchup_label")
])

matchup_board
```

```python
matchup_board = matchup_board.with_columns([
    pl.when(pl.col("position") == "RB")
      .then(pl.col("rush_def_rank"))
      .otherwise(pl.col("pass_def_rank"))
      .alias("key_def_rank")
])

matchup_board
```
